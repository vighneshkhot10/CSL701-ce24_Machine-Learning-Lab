```python
# Task 1: Import the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 2: Load train.csv into a Pandas DataFrame
df = pd.read_csv("Titanic-Dataset(1).csv")

# Display the first 5 rows to verify the dataset loaded successfully
df.head()
```

## Phase 3: Data Inspection

### Task 3: Display first 10 rows using `df.head(10)`.

```python
display(df.head(10))
```

### Task 4: Display last 5 rows using `df.tail(5)`.

```python
display(df.tail(5))
```

### Task 5: Check dataset dimensions using `df.shape` (Expected: 891 x 12).

```python
print(f"Dataset dimensions: {df.shape}")
```

### Task 6: Display dataset information using `df.info()`.

```python
df.info()
```

### Task 7: Generate summary statistics using `df.describe()`; record mean, minimum and maximum values of Age and Fare.

```python
summary_stats = df.describe()
display(summary_stats)

age_mean = summary_stats.loc['mean', 'Age']
age_min = summary_stats.loc['min', 'Age']
age_max = summary_stats.loc['max', 'Age']

fare_mean = summary_stats.loc['mean', 'Fare']
fare_min = summary_stats.loc['min', 'Fare']
fare_max = summary_stats.loc['max', 'Fare']

print(f"\nAge: Mean = {age_mean:.2f}, Minimum = {age_min:.2f}, Maximum = {age_max:.2f}")
print(f"Fare: Mean = {fare_mean:.2f}, Minimum = {fare_min:.2f}, Maximum = {fare_max:.2f}")
```

### Task 8: Identify missing values using `df.isnull().sum()`.

```python
print("Missing values per column:")
print(df.isnull().sum())
```

### Task 9: Fill missing Age values with the median.

```python
median_age = df['Age'].median()
df['Age'].fillna(median_age, inplace=True)
print(f"Missing values in 'Age' after filling: {df['Age'].isnull().sum()}")
```

### Task 10: Filter surviving passengers (Survived == 1).

```python
survived_passengers = df[df['Survived'] == 1]
display(survived_passengers.head())
```

### Task 11: Convert Fare column to a NumPy array.

```python
fare_array = df['Fare'].to_numpy()
print(f"Type of fare_array: {type(fare_array)}")
print(f"First 5 elements of fare_array: {fare_array[:5]}")
```

### Task 12: Calculate maximum and average Fare using `np.max()` and `np.mean()`.

```python
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)
print(f"Maximum Fare: {max_fare:.2f}")
print(f"Average Fare: {avg_fare:.2f}")
```

### Task 13: Plot histogram of Age.

```python
# Plot histogram of Age
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Task 14: Plot bar chart of Survival Count.

```python
# Plot bar chart of Survival Count
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df)
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Task 15: Plot scatter graph of Age vs Fare.

```python
# Plot scatter graph of Age vs Fare
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df, hue='Survived', palette='viridis', alpha=0.7)
plt.title('Age vs Fare (Colored by Survival Status)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
